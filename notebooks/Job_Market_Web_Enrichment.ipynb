```python
from selenium import webdriver
from selenium.webdriver.edge.options import Options
from selenium.webdriver.common.by import By

import pandas as pd
import re
import time

options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Edge(options=options)

print("Driver started")
```

```python
def extract_figma_job(url):
    
    result = {'location_clean': None,'salary_currency': None, 'salary_min': None,
        'salary_max': None, 'education': None,'content_clean': None, 'page_type': None }
    
    try:
        driver.get(url)
        time.sleep(3)
        
        # Get page information
        page_title = driver.title
        page_text = driver.find_element(By.TAG_NAME, "body").text
        
        result['content_clean'] = page_text
        
        title_lower = page_title.lower()
        text_lower = page_text.lower()
        

        # 1. CHECK 404 / UNAVAILABLE PAGE

        unavailable_patterns = ['404','page not found', 'job not found',
            'this job is no longer available','position is no longer available']
        
        if any(pattern in text_lower for pattern in unavailable_patterns):
            result['page_type'] = '404'
            result['content_clean'] = None
            return result
        

        # 2. DETECT GENERAL FIGMA CAREERS PAGE

        general_page_indicators = [ 'careers at figma','see open positions',
            'job openings','our values and culture']
        
        general_score = sum(
            indicator in text_lower
            for indicator in general_page_indicators
        )
        

        # 3. DETECT INDIVIDUAL JOB PAGE

        job_page_indicators = ['apply for this job','apply', "what you'll do",
            'annual base salary range:','qualifications', 'requirements' ]
        
        job_score = sum(indicator in text_lower for indicator in job_page_indicators)
        
        # If strong evidence of general page and
        # insufficient evidence of individual job
        if general_score >= 2 and job_score < 2:
            result['page_type'] = 'general_page'
            result['content_clean'] = None
            return result
        
        # If enough evidence of an individual job
        if job_score >= 2:
            result['page_type'] = 'job_listing'
        else:
            result['page_type'] = 'general_page'
            result['content_clean'] = None
            return result
        

        # 4. EXTRACT LOCATION

        lines = [ line.strip()for line in page_text.split('\n') if line.strip()]
        
        # Find "Back to jobs"
        back_to_jobs_index = None
        
        for i, line in enumerate(lines):
            if line.lower() == 'back to jobs':
                back_to_jobs_index = i
                break
        
        if back_to_jobs_index is not None:
            
            # Usually:
            # Back to jobs
            # Job title
            # Location
            
            if back_to_jobs_index + 2 < len(lines):
                possible_location = lines[back_to_jobs_index + 2]
                
                # Don't accidentally take "Apply"
                if possible_location.lower() != 'apply':
                    result['location_clean'] = possible_location
        

        # 5. EXTRACT SALARY

        salary_patterns = [
            r'Annual Base Salary Range:\s*'
            r'\$([\d,]+)\s*-\s*\$([\d,]+)\s*'
            r'(USD|CAD|EUR|GBP)',
            
            r'Base Salary Range:\s*'
            r'\$([\d,]+)\s*-\s*\$([\d,]+)\s*'
            r'(USD|CAD|EUR|GBP)'
        ]
        
        for pattern in salary_patterns:
            
            salary_match = re.search(pattern, page_text, re.IGNORECASE )
            
            if salary_match:
                
                result['salary_min'] = float(salary_match.group(1).replace(',', ''))
                
                result['salary_max'] = float( salary_match.group(2).replace(',', ''))
                
                result['salary_currency'] = (salary_match.group(3).upper())
                
                break
        

        # 6. EXTRACT EDUCATION

        education_patterns = [ r"((?:Bachelor['’]?s|Master['’]?s|Ph\.?D\.?)[^\n]*)"]
        
        for pattern in education_patterns:
            
            education_match = re.search(pattern,page_text, re.IGNORECASE )
            
            if education_match:
                
                education_text = education_match.group(1).strip()
                
                result['education'] = education_text
                
                break
        
        return result
        
    except Exception as e:
        
        print(
            f"ERROR: {type(e).__name__}: {e}"
        )
        
        result['page_type'] = 'error'
        result['content_clean'] = None
        
        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_cockroach_job(url):

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:
        driver.get(url)
        time.sleep(5)

        # Find Greenhouse iframe
        iframe = driver.find_element(
            By.CSS_SELECTOR,
            'iframe[src*="job-boards.greenhouse.io/embed/job_app"]'
        )

        # Switch into Greenhouse iframe
        driver.switch_to.frame(iframe)
        time.sleep(2)

        page_text = driver.find_element(By.TAG_NAME, "body").text


        # Check whether this is a job
        text_lower = page_text.lower()

        job_indicators = ['back to jobs','apply here','the role', 'you will','you have' ]

        job_score = sum(indicator in text_lower for indicator in job_indicators)

        if job_score >= 2:
            result['page_type'] = 'job_listing'
        else:
            result['page_type'] = 'general_page'
            driver.switch_to.default_content()
            return result


        # Extract location

        lines = [line.strip()
            for line in page_text.split('\n')
            if line.strip()]

        back_to_jobs_index = None

        for i, line in enumerate(lines):
            if line.lower() == 'back to jobs':
                back_to_jobs_index = i
                break

        if back_to_jobs_index is not None:

            if back_to_jobs_index + 1 < len(lines):

                possible_location = lines[back_to_jobs_index + 2]

                if (
                    possible_location.lower() not in
                    ['apply here!', 'apply here']
                ):
                    result['location_clean'] = possible_location


        # Extract salary

        salary_patterns = [

            # $145,000 - $190,000 USD
            r'\$([\d,]+)\s*-\s*\$([\d,]+)\s*(USD|CAD|EUR|GBP)',

            # $145,000–$190,000 USD
            r'\$([\d,]+)\s*[–-]\s*\$([\d,]+)\s*(USD|CAD|EUR|GBP)',

            # 145,000 - 190,000 USD
            r'([\d,]+)\s*-\s*([\d,]+)\s*(USD|CAD|EUR|GBP)'
        ]

        for pattern in salary_patterns:

            salary_match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if salary_match:

                result['salary_min'] = float(
                    salary_match.group(1).replace(',', '')
                )

                result['salary_max'] = float(
                    salary_match.group(2).replace(',', '')
                )

                result['salary_currency'] = (
                    salary_match.group(3).upper()
                )

                break


        # Extract education

        education_patterns = [

            r"((?:Bachelor['’]?s|Master['’]?s|Ph\.?D\.?)[^\n.]*)",

            r"((?:Bachelor['’]?s|Master['’]?s|Ph\.?D\.?)[^.\n]*degree[^.\n]*)"
        ]

        for pattern in education_patterns:

            education_match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if education_match:

                education = education_match.group(1).strip()

                # Avoid extremely long accidental matches
                if len(education) <= 300:
                    result['education'] = education

                break

        driver.switch_to.default_content()

        return result

    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        try:
            driver.switch_to.default_content()
        except:
            pass

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


def extract_cockroach_job(url):

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:
        driver.get(url)

        wait = WebDriverWait(driver, 15)

        # Wait for the Greenhouse iframe
        iframe = wait.until(
            EC.presence_of_element_located(
                (
                    By.CSS_SELECTOR,
                    'iframe[src*="job-boards.greenhouse.io/embed/job_app"]'
                )
            )
        )

        # Switch into iframe
        driver.switch_to.frame(iframe)

        # Wait until body exists inside iframe
        wait.until(
            EC.presence_of_element_located(
                (By.TAG_NAME, "body")
            )
        )

        # Give Greenhouse a little extra time to render content
        time.sleep(2)

        page_text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        text_lower = page_text.lower()


        # Detect individual job vs listing page


        job_indicators = ['back to jobs','apply here','the role','you will','you have']

        job_score = sum(
            indicator in text_lower
            for indicator in job_indicators
        )

        listing_indicators = ['create a job alert','applications (general)',
            'search','jobs']

        listing_score = sum(
            indicator in text_lower
            for indicator in listing_indicators
        )

        if job_score >= 2:
            result['page_type'] = 'individual_job'

        elif listing_score >= 2:
            result['page_type'] = 'job_listing_page'

            driver.switch_to.default_content()
            return result

        else:
            result['page_type'] = 'unknown_page'

            driver.switch_to.default_content()
            return result

 
        # Extract location


        lines = [
            line.strip()
            for line in page_text.split('\n')
            if line.strip()
        ]

        back_to_jobs_index = None

        for i, line in enumerate(lines):

            if line.lower() == 'back to jobs':
                back_to_jobs_index = i
                break

        if back_to_jobs_index is not None:

            location_index = back_to_jobs_index + 2

            if location_index < len(lines):

                possible_location = lines[location_index]

                if possible_location.lower() not in [
                    'apply here',
                    'apply here!'
                ]:
                    result['location_clean'] = possible_location


        # Extract salary


        salary_patterns = [
            r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)\s*(USD|CAD|EUR|GBP)',
            r'([\d,]+)\s*[-–]\s*([\d,]+)\s*(USD|CAD|EUR|GBP)'
        ]

        for pattern in salary_patterns:

            match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if match:

                result['salary_min'] = float(
                    match.group(1).replace(',', '')
                )

                result['salary_max'] = float(
                    match.group(2).replace(',', '')
                )

                result['salary_currency'] = (
                    match.group(3).upper()
                )

                break


        # Extract explicit education


        education_patterns = [
            r"((?:Bachelor['’]?s|Master['’]?s|Ph\.?D\.?)[^.\n]*)"
        ]

        for pattern in education_patterns:

            match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if match:

                education = match.group(1).strip()

                if len(education) <= 300:
                    result['education'] = education

                break

        driver.switch_to.default_content()

        return result

    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        try:
            driver.switch_to.default_content()
        except:
            pass

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_brex_job(url):

    result = { 'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:
        driver.get(url)
        time.sleep(4)

        page_text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        text_lower = page_text.lower()


        # Detect 404

        if (
            '404 page not found' in text_lower
            or "sorry, we can’t seem to find that page" in text_lower
            or "sorry, we can't seem to find that page" in text_lower
        ):
            result['page_type'] = '404'
            return result


        # Detect individual job page


        job_indicators = [
            'why join us',
            'what you’ll do',
            "what you'll do",
            'requirements',
            'compensation'
        ]

        job_score = sum(
            indicator in text_lower
            for indicator in job_indicators
        )

        if job_score >= 2:
            result['page_type'] = 'individual_job'
        else:
            result['page_type'] = 'unknown_page'
            return result


        # Extract location


        lines = [
            line.strip()
            for line in page_text.split('\n')
            if line.strip()
        ]

        # Brex structure:
        #
        # Careers
        # Job title
        # Job title
        # Location

        for i, line in enumerate(lines):

            if line.lower() == 'careers':

                if i + 3 < len(lines):

                    possible_location = lines[i + 3]

                    if possible_location:
                        result['location_clean'] = possible_location

                    break


        # Extract salary


        salary_patterns = [

            # $240,000 - $300,000
            r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)',

            # $240,000–$300,000
            r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)'
        ]

        for pattern in salary_patterns:

            match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if match:

                result['salary_min'] = float(
                    match.group(1).replace(',', '')
                )

                result['salary_max'] = float(
                    match.group(2).replace(',', '')
                )

                # Brex salary is explicitly in dollars.
                # We'll verify USD from surrounding text.
                salary_context = page_text[
                    max(0, match.start() - 200):
                    min(len(page_text), match.end() + 200)
                ].lower()

                if 'usd' in salary_context:
                    result['salary_currency'] = 'USD'
                elif '$' in match.group(0):
                    # Brex compensation page uses $
                    # but only assign USD when context
                    # supports it.
                    result['salary_currency'] = 'USD'

                break


        # Extract explicit education


        education_patterns = [
            r"((?:Bachelor['’]?s|Master['’]?s|Ph\.?D\.?)[^.\n]*)"
        ]

        for pattern in education_patterns:

            match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if match:

                education = match.group(1).strip()

                if len(education) <= 300:
                    result['education'] = education

                break

        return result

    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_vercel_job(url):

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:
        # Open job page
        driver.get(url)
        time.sleep(3)

        page_title = driver.title
        page_text = driver.find_element(
            By.TAG_NAME, "body"
        ).text

        text_lower = page_text.lower()


        # 1. Check for 404 / unavailable page


        unavailable_patterns = [
            '404',
            'page not found',
            'job not found',
            'this job is no longer available',
            'position is no longer available'
        ]

        if any(
            pattern in text_lower
            for pattern in unavailable_patterns
        ):
            result['page_type'] = '404'
            return result


        # 2. Check whether this is an individual job page


        job_indicators = [
            'back to jobs',
            'apply',
            'about vercel:',
            'about the role:',
            'what you will do:',
            "what you'll do:",
            'requirements',
            'benefits'
        ]

        job_score = sum(
            indicator in text_lower
            for indicator in job_indicators
        )

        if job_score >= 3:
            result['page_type'] = 'individual_job'
        else:
            result['page_type'] = 'unknown_page'
            return result


        # 3. Extract location
        lines = [
            line.strip()
            for line in page_text.split('\n')
            if line.strip()
        ]

        back_to_jobs_index = None

        for i, line in enumerate(lines):
            if line.lower() == 'back to jobs':
                back_to_jobs_index = i
                break

        if back_to_jobs_index is not None:

            location_index = back_to_jobs_index + 2

            if location_index < len(lines):

                possible_location = lines[location_index]

                if (
                    possible_location.lower() != 'apply'
                    and possible_location.lower() != 'back to jobs'
                    and len(possible_location) <= 300
                ):
                    result['location_clean'] = possible_location


        # 4. Extract salary
        salary_patterns = [

            # Example:
            # 388,000 - 524,000 AUD OTE
            r'([\d,]+)\s*[-–]\s*([\d,]+)\s*'
            r'(AUD|USD|CAD|EUR|GBP|SGD|INR|JPY|CHF|NZD)'
            r'(?:\s+OTE)?',

            # Example:
            # $200,000 - $300,000 USD
            r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)\s*'
            r'(USD|AUD|CAD|EUR|GBP|SGD|INR|JPY|CHF|NZD)',

            # Example:
            # 200,000 - 300,000 USD
            r'([\d,]+)\s*[-–]\s*([\d,]+)\s*'
            r'(USD|AUD|CAD|EUR|GBP|SGD|INR|JPY|CHF|NZD)'
        ]

        salary_found = False

        for pattern in salary_patterns:

            match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )

            if match:

                result['salary_min'] = float(
                    match.group(1).replace(',', '')
                )

                result['salary_max'] = float(
                    match.group(2).replace(',', '')
                )

                if len(match.groups()) >= 3:

                    result['salary_currency'] = (
                        match.group(3).upper()
                    )

                salary_found = True
                break

        # --------------------------------------------------
        # 5. Dollar range without explicit currency
        # --------------------------------------------------
        # Extract salary numbers but DO NOT assume USD.
        # --------------------------------------------------

        if not salary_found:

            dollar_match = re.search(
                r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)',
                page_text,
                re.IGNORECASE
            )

            if dollar_match:

                result['salary_min'] = float(
                    dollar_match.group(1).replace(',', '')
                )

                result['salary_max'] = float(
                    dollar_match.group(2).replace(',', '')
                )

                # Currency intentionally remains None.


        # 6. Extract explicit education requirement

        education_patterns = [

            r"\b(Bachelor['’]s"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})",

            r"\b(Master['’]s"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})",

            r"\b(Ph\.?\s*D\.?"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})",

            r"\b(Doctorate"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})"
        ]

        for pattern in education_patterns:

            matches = re.finditer(
                pattern,
                page_text,
                re.IGNORECASE
            )

            for match in matches:

                education = match.group(1).strip()

                education_lower = education.lower()

                # Reject false matches such as:
                # "master your craft"
                # "master the..."
                # "mastery..."
                false_positive_patterns = [
                    'master your',
                    'master the',
                    'mastering ',
                    'mastery',
                    'masterfully'
                ]

                if any(
                    bad in education_lower
                    for bad in false_positive_patterns
                ):
                    continue

                if 5 <= len(education) <= 300:

                    result['education'] = education
                    break

            if result['education'] is not None:
                break

        return result

    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_checkr_job(url):

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:


        # 1. Open page

        driver.get(url)
        time.sleep(3)

        page_text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        text_lower = page_text.lower()


        # 2. Check for unavailable / 404 page

        unavailable_patterns = ['404', 'page not found',
            'job not found','this job is no longer available',
            'position is no longer available']

        if any(
            pattern in text_lower
            for pattern in unavailable_patterns
        ):

            result['page_type'] = '404'

            return result



        # 3. Check individual job page

        job_indicators = ['back to jobs','apply','about checkr',
            'what you’ll do',"what you'll do",'what we’re looking for',
            "what we're looking for",'pay transparency disclosure','what we offer']

        job_score = sum(indicator in text_lower for indicator in job_indicators)

        if job_score >= 3:

            result['page_type'] = 'individual_job'

        else:

            result['page_type'] = 'unknown_page'

            return result



        # 4. Extract location


        lines = [ line.strip()
            for line in page_text.split('\n')
            if line.strip()]

        back_to_jobs_index = None

        for i, line in enumerate(lines):

            if line.lower() == 'back to jobs':

                back_to_jobs_index = i
                break


        if back_to_jobs_index is not None:

            location_index = back_to_jobs_index + 2

            if location_index < len(lines):

                possible_location = lines[
                    location_index
                ]

                if (
                    possible_location.lower() != 'apply'
                    and possible_location.lower() != 'back to jobs'
                    and len(possible_location) <= 300
                ):

                    result['location_clean'] = (
                        possible_location
                    )



        # 5. Extract salary

        salary_patterns = [

            # Example:
            # $169,000 - $199,000 USD
            r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)\s*'
            r'(USD|AUD|CAD|EUR|GBP|SGD|INR|JPY|CHF|NZD)',

            # Example:
            # 169,000 - 199,000 USD
            r'([\d,]+)\s*[-–]\s*([\d,]+)\s*'
            r'(USD|AUD|CAD|EUR|GBP|SGD|INR|JPY|CHF|NZD)',

            # Example:
            # 169,000 - 199,000 USD OTE
            r'([\d,]+)\s*[-–]\s*([\d,]+)\s*'
            r'(USD|AUD|CAD|EUR|GBP|SGD|INR|JPY|CHF|NZD)'
            r'(?:\s+OTE)?'
        ]


        salary_found = False

        for pattern in salary_patterns:

            match = re.search(pattern,page_text, re.IGNORECASE)

            if match:

                result['salary_min'] = float(
                    match.group(1).replace(',', ''))

                result['salary_max'] = float(
                    match.group(2).replace(',', '')
                )

                if len(match.groups()) >= 3:

                    result['salary_currency'] = (
                        match.group(3).upper()
                    )

                salary_found = True

                break



        # 6. Dollar range without explicit currency
 

        if not salary_found:

            dollar_match = re.search(
                r'\$([\d,]+)\s*[-–]\s*\$([\d,]+)',
                page_text,
                re.IGNORECASE
            )

            if dollar_match:

                result['salary_min'] = float(
                    dollar_match.group(1).replace(',', '')
                )

                result['salary_max'] = float(
                    dollar_match.group(2).replace(',', '')
                )

                # Do NOT assume USD.



        # 7. Extract explicit education requirement


        education_patterns = [

            r"\b(Bachelor['’]s"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})",

            r"\b(Master['’]s"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})",

            r"\b(Ph\.?\s*D\.?"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})",

            r"\b(Doctorate"
            r"(?:\s+degree)?"
            r"[^.\n]{0,250})"
        ]


        for pattern in education_patterns:

            matches = re.finditer(
                pattern,
                page_text,
                re.IGNORECASE
            )

            for match in matches:

                education = match.group(1).strip()

                education_lower = education.lower()


                # Reject obvious false positives
                false_positive_patterns = ['master your','master the',
                    'mastering ', 'mastery', 'masterfully']

                if any(
                    bad in education_lower
                    for bad in false_positive_patterns
                ):

                    continue


                if 5 <= len(education) <= 300:

                    result['education'] = education

                    break


            if result['education'] is not None:

                break


        return result


    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_gusto_job(url):

    result = {
        'location_clean': None,
        'salary_ranges': [],
        'education': None,
        'page_type': None
    }

    try:

        # OPEN PAGE

        driver.get(url)
        time.sleep(3)

        page_text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        text_lower = page_text.lower()


        # CHECK 404 / UNAVAILABLE PAGE

        unavailable_patterns = ['404',
            'page not found',
            'job not found',
            'this job is no longer available',
            'position is no longer available' ]

        if any(
            pattern in text_lower
            for pattern in unavailable_patterns
        ):
            result['page_type'] = '404'
            return result


        # IDENTIFY INDIVIDUAL JOB PAGE

        job_indicators = ['back to jobs','apply','about gusto',
            'about the role',"here's what you'll do", "here's what we're looking for",
            'what we’re looking for','cash compensation','total rewards' ]

        job_score = sum(
            indicator in text_lower
            for indicator in job_indicators
        )

        if job_score >= 3:
            result['page_type'] = 'individual_job'
        else:
            result['page_type'] = 'unknown_page'
            return result


        # EXTRACT LOCATION
  
        lines = [ line.strip()
            for line in page_text.split('\n')
            if line.strip() ]

        back_to_jobs_index = None

        for i, line in enumerate(lines):

            if line.lower() == 'back to jobs':
                back_to_jobs_index = i
                break

        if back_to_jobs_index is not None:

            location_index = back_to_jobs_index + 2

            if location_index < len(lines):

                possible_location = lines[location_index]

                if (
                    possible_location.lower() != 'apply'
                    and possible_location.lower() != 'back to jobs'
                    and len(possible_location) <= 300
                ):
                    result['location_clean'] = possible_location


        # EXTRACT SALARY DISCLOSURE

        salary_sentences = [
            line.strip()
            for line in lines
            if (
                'cash compensation' in line.lower()
                or 'compensation amount' in line.lower()
            )
        ]

        # ---------------------------------------------------------
        # FORMAT A
        #
        # $153,000 to $195,000 USD for Denver...
        # $180,815 to $230,000 USD for San Francisco...
        # $146,940-$185,000 CAD for Toronto
        # ---------------------------------------------------------
        pattern_with_currency = re.compile(
            r'\$([\d,]+)'
            r'\s*(?:to|[-–])\s*'
            r'\$?([\d,]+)'
            r'\s*'
            r'(USD|CAD|AUD|EUR|GBP|SGD|INR|JPY|CHF|NZD)'
            r'\s*'
            r'(?:for|in)\s*'
            r'([^$]+?)'
            r'(?=\s+and\s+\$|\.\s*Final offer)',
            re.IGNORECASE
        )

        # ---------------------------------------------------------
        # FORMAT B
        #
        # $189,000–$236,000 per year in Denver...
        # $223,000–$278,000 per year in San Francisco...
        #
        # Currency is not immediately after the range.
        # Gusto's disclosure is USD in these examples.
        # ---------------------------------------------------------
        pattern_without_currency = re.compile(
            r'\$([\d,]+)'
            r'\s*[–-]\s*'
            r'\$([\d,]+)'
            r'(?:\s+per\s+year|\s*/yr)?'
            r'\s+(?:in|for)\s+'
            r'([^$]+?)'
            r'(?=,\s*and\s+\$|\s+and\s+\$|\.\s*Stock|\.\s*Final offer)',
            re.IGNORECASE
        )

        for salary_text in salary_sentences:


            # FIRST: ranges where currency is explicitly present

            matches = pattern_with_currency.findall( salary_text )

            for match in matches:

                salary_record = {
                    'salary_currency': match[2].upper(),
                    'salary_min': float(
                        match[0].replace(',', '')
                    ),
                    'salary_max': float(
                        match[1].replace(',', '')
                    ),
                    'salary_location': match[3].strip()
                }

                if salary_record not in result['salary_ranges']:
                    result['salary_ranges'].append(
                        salary_record
                    )


            # SECOND: ranges where currency isn't beside range

            matches = pattern_without_currency.findall(
                salary_text
            )

            for match in matches:

                salary_record = {
                    'salary_currency': 'USD',
                    'salary_min': float(
                        match[0].replace(',', '')
                    ),
                    'salary_max': float(
                        match[1].replace(',', '')
                    ),
                    'salary_location': match[2].strip()
                }

                if salary_record not in result['salary_ranges']:
                    result['salary_ranges'].append(
                        salary_record
                    )


        # EXTRACT EDUCATION

        education_patterns = [
            r"\b(Bachelor['’]s(?:\s+degree)?[^.\n]{0,250})",
            r"\b(Master['’]s(?:\s+degree)?[^.\n]{0,250})",
            r"\b(Ph\.?\s*D\.?(?:\s+degree)?[^.\n]{0,250})",
            r"\b(Doctorate(?:\s+degree)?[^.\n]{0,250})"
        ]

        false_positive_patterns = ['master your','master the','mastering ','mastery','masterfully' ]

        for pattern in education_patterns:

            matches = re.finditer( pattern,page_text,re.IGNORECASE)

            for match in matches:

                education = match.group(1).strip()

                education_lower = education.lower()

                if any(
                    bad in education_lower
                    for bad in false_positive_patterns
                ):
                    continue

                if 5 <= len(education) <= 300:

                    result['education'] = education
                    break

            if result['education'] is not None:
                break

        return result

    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_datadog_job(url):

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:
        driver.get(url)
        time.sleep(3)

        page_text = driver.find_element(
            By.TAG_NAME, "body"
        ).text

        text_lower = page_text.lower()


        # 1. Check unavailable pages


        unavailable_patterns = [
            '404',
            'page not found',
            'job not found',
            'this job is no longer available',
            'position is no longer available'
        ]

        if any(
            pattern in text_lower
            for pattern in unavailable_patterns
        ):
            result['page_type'] = '404'
            return result


        # 2. Identify individual job page

        job_indicators = ["what you'll do","who you are", "benefits and growth",
            "about datadog","datadog offers", "requisition id", "salary" ]

        job_score = sum(
            indicator in text_lower
            for indicator in job_indicators
        )

        if job_score >= 3:
            result['page_type'] = 'individual_job'
        else:
            result['page_type'] = 'unknown_page'
            return result


        # 3. Extract location


        lines = [line.strip()
            for line in page_text.split('\n')
            if line.strip() ]

        # Datadog normally places location near the top,
        # immediately after the job title.

        location_patterns = [
            r'^.+,\s*.+,\s*USA$',
            r'^.+,\s*.+,\s*US$',
            r'^.+,\s*.+,\s*Canada$',
            r'^.+,\s*.+,\s*UK$',
            r'^.+,\s*.+,\s*United Kingdom$',
            r'^.+,\s*.+,\s*Australia$',
            r'^.+,\s*.+,\s*Germany$',
            r'^.+,\s*.+,\s*France$',
            r'^.+,\s*.+,\s*Netherlands$',
            r'^.+,\s*.+,\s*Singapore$',
            r'^.+,\s*.+,\s*Japan$'
        ]

        # Search only the beginning of the page so that
        # locations mentioned later in the job description
        # aren't incorrectly selected.

        for line in lines[:20]:

            for pattern in location_patterns:

                if re.match(
                    pattern,
                    line,
                    re.IGNORECASE
                ):
                    result['location_clean'] = line
                    break

            if result['location_clean'] is not None:
                break


        # 4. Extract salary

        # Examples:
        # $170,000—$227,000 USD
        # $150,000-$200,000 USD
        # $150,000–$200,000 USD

        salary_patterns = [r'\$([\d,]+)\s*[—–-]\s*\$([\d,]+)\s*(USD|CAD|AUD|EUR|GBP|SGD|INR|JPY|CHF|NZD)',
            r'\$([\d,]+)\s*(?:to)\s*\$([\d,]+)\s*(USD|CAD|AUD|EUR|GBP|SGD|INR|JPY|CHF|NZD)'  ]

        for pattern in salary_patterns:

            salary_match = re.search(  pattern, page_text, re.IGNORECASE )

            if salary_match:

                result['salary_min'] = float( salary_match.group(1).replace(',', '') )

                result['salary_max'] = float(salary_match.group(2).replace(',', '') )

                result['salary_currency'] = (salary_match.group(3).upper())
                break


        # 5. Extract education

        education_patterns = [

            # BA/BS
            r'\b(BA/BS[^.\n]{0,250})',

            # Bachelor's
            r"\b(Bachelor['’]s(?:\s+degree)?[^.\n]{0,250})",

            # Master's
            r"\b(Master['’]s(?:\s+degree)?[^.\n]{0,250})",

            # PhD
            r'\b(Ph\.?\s*D\.?(?:\s+degree)?[^.\n]{0,250})',

            # Doctorate
            r'\b(Doctorate(?:\s+degree)?[^.\n]{0,250})',

            # "degree in Computer Science..."
            r'\b(degree\s+in\s+[A-Za-z][^.\n]{0,250})'
        ]

        false_positive_patterns = [ 'master your','master the', 'mastering ',  'mastery','masterfully']

        for pattern in education_patterns:

            matches = re.finditer( pattern,page_text, re.IGNORECASE)

            for match in matches:

                education = match.group(1).strip()
                education_lower = education.lower()

                if any(
                    bad in education_lower
                    for bad in false_positive_patterns
                ):
                    continue

                if 5 <= len(education) <= 300:

                    result['education'] = education

                    break

            if result['education'] is not None:
                break

        return result

    except Exception as e:

        print( f"ERROR for {url}: " f"{type(e).__name__}: {e}" )

        result['page_type'] = 'error'

        return result
```

```python
import re
import time
from selenium.webdriver.common.by import By


def extract_mongodb_job(url):

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None
    }

    try:
        driver.get(url)
        time.sleep(3)

        page_text = driver.find_element(
            By.TAG_NAME,
            "body"
        ).text

        text_lower = page_text.lower()


        # Check unavailable pages

        unavailable_patterns = [ '404','page not found', 'job not found',
            'this job is no longer available','position is no longer available' ]

        if any(
            pattern in text_lower
            for pattern in unavailable_patterns
        ):
            result['page_type'] = '404'
            return result


        # Identify individual job page

        job_indicators = [
            "what you'll do",
            "who you are",
            "about mongodb",
            "req id",
            "salary range",
            "base salary range",
            "success measures"
        ]

        job_score = sum( indicator in text_lower for indicator in job_indicators  )

        if job_score >= 3:
            result['page_type'] = 'individual_job'
        else:
            result['page_type'] = 'unknown_page'
            return result


        # LOCATION

        lines = [line.strip() for line in page_text.split('\n') if line.strip() ]

        # Explicit remote-location statements
        remote_patterns = [
            r'this role can be based remotely in the united states',
            r'can be based remotely in the united states',
            r'remote in the united states',
            r'remotely in the united states'
        ]

        for pattern in remote_patterns:

            if re.search( pattern, page_text,
                re.IGNORECASE
            ):
                result['location_clean'] = (
                    'Remote, United States'
                )
                break

        # Normal location patterns
        if result['location_clean'] is None:

            location_patterns = [
                r'^Remote\s*[-,]?\s*.+$',
                r'^.+,\s*United States$',
                r'^.+,\s*USA$',
                r'^.+,\s*Canada$',
                r'^.+,\s*United Kingdom$',
                r'^.+,\s*UK$',
                r'^.+,\s*Australia$',
                r'^.+,\s*Germany$',
                r'^.+,\s*France$',
                r'^.+,\s*Singapore$'
            ]

            for line in lines[:40]:

                for pattern in location_patterns:

                    if re.match(
                        pattern,
                        line,
                        re.IGNORECASE
                    ):
                        result['location_clean'] = line
                        break

                if result['location_clean'] is not None:
                    break


        # SALARY

        salary_patterns = [

            # Example:
            # $192,800—$241,000 USD
            r'\$([\d,]+)\s*[—–-]\s*\$?([\d,]+)\s*'
            r'(USD|CAD|AUD|EUR|GBP|SGD|INR|JPY|CHF|NZD)',

            # Example:
            # $192,800 - $241,000
            r'\$([\d,]+)\s*[—–-]\s*\$?([\d,]+)'
        ]

        for pattern in salary_patterns:

            salary_match = re.search( pattern,page_text, re.IGNORECASE )

            if salary_match:

                result['salary_min'] = float(salary_match.group(1).replace(',', ''))

                result['salary_max'] = float(salary_match.group(2).replace(',', ''))

                # Currency only when explicitly stated
                if salary_match.lastindex >= 3:

                    result['salary_currency'] = (salary_match.group(3).upper())

                break


        # EDUCATION

        education_patterns = [
            r"\bBachelor['’]s(?:\s+degree)?[^.\n]{0,250}",
            r"\bMaster['’]s(?:\s+degree)?[^.\n]{0,250}",
            r'\bBA/BS[^.\n]{0,250}',
            r'\bB\.?Tech\.?[^.\n]{0,250}',
            r'\bM\.?Tech\.?[^.\n]{0,250}',
            r'\bMBA[^.\n]{0,250}',
            r'\bPh\.?\s*D\.?(?:\s+degree)?[^.\n]{0,250}',
            r'\bdegree in [A-Za-z][^.\n]{0,250}'
        ]

        false_positive_patterns = ['master your','master the', 'mastering', 'mastery' ]

        for pattern in education_patterns:

            matches = re.finditer( pattern, page_text, re.IGNORECASE)

            for match in matches:

                education = match.group(0).strip()

                if any(
                    bad in education.lower()
                    for bad in false_positive_patterns
                ):
                    continue

                if 5 <= len(education) <= 300:

                    result['education'] = education
                    break

            if result['education'] is not None:
                break

        return result

    except Exception as e:

        print(
            f"ERROR for {url}: "
            f"{type(e).__name__}: {e}"
        )

        result['page_type'] = 'error'

        return result
```
