- Source-specific fields
- Information not provided in the original job posting

- Location information not explicitly available

```python
import re

def extract_figma_job(url):
    
    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'content_clean': None,
        'page_type': None
    }
    
    try:
        driver.get(url)
        time.sleep(3)
        
        page_text = driver.find_element(By.TAG_NAME, "body").text
        
        # Store full page content
        result['content_clean'] = page_text
        
        # -------------------------
        # PAGE TYPE
        # -------------------------
        if 'Annual Base Salary Range:' in page_text:
            result['page_type'] = 'job_listing'
        else:
            result['page_type'] = 'general_page'
            return result
        
        # -------------------------
        # LOCATION
        # -------------------------
        lines = [line.strip() for line in page_text.split('\n') if line.strip()]
        
        # Usually:
        # line 1 = Back to jobs
        # line 2 = Job title
        # line 3 = Location
        if len(lines) >= 3:
            result['location_clean'] = lines[2]
        
        # -------------------------
        # SALARY
        # -------------------------
        salary_match = re.search(
            r'Annual Base Salary Range:\s*'
            r'\$([\d,]+)\s*-\s*\$([\d,]+)\s*'
            r'(USD|CAD|EUR|GBP)',
            page_text,
            re.IGNORECASE
        )
        
        if salary_match:
            result['salary_min'] = float(
                salary_match.group(1).replace(',', '')
            )
            
            result['salary_max'] = float(
                salary_match.group(2).replace(',', '')
            )
            
            result['salary_currency'] = salary_match.group(3).upper()
        
        # -------------------------
        # EDUCATION
        # -------------------------
        education_patterns = [
            r"(Bachelor['’]?s[^.\n]*)",
            r"(Master['’]?s[^.\n]*)",
            r"(Ph\.?D\.?[^.\n]*)"
        ]
        
        for pattern in education_patterns:
            education_match = re.search(
                pattern,
                page_text,
                re.IGNORECASE
            )
            
            if education_match:
                result['education'] = education_match.group(1).strip()
                break
        
        return result
    
    except Exception as e:
        print(f"Error: {type(e).__name__}: {e}")
        result['page_type'] = 'error'
        return result
```
