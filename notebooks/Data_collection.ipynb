```python
APP_ID="aabebcdf"
APP_KEY="84ac0e372e46c2b2a66f9dcb4f6f204e"

roles = [
    "software engineer",
    "data scientist",
    "data analyst",
    "machine learning engineer",
    "backend developer",
    "frontend developer",
    "full stack developer",
    "python developer",
    "java developer",
    "devops engineer",
    "cloud engineer",
    "cyber security",
    "business analyst"
]
max_pages=20
all_dataframes=[]
def fetch_data(roles,page):
    url = f"https://api.adzuna.com/v1/api/jobs/in/search/{page}"
    params={
        "app_id":APP_ID,
        "app_key":APP_KEY,
        "results_per_page":20,
        "what":roles
    }
    response=requests.get(url, params=params)
    if response.status_code != 200:
       print(f"Error: {response.status_code}")
       return pd.DataFrame()
    data=response.json()
    if "results"  not in data:
        return pd.DataFrame()
    return pd.DataFrame(data['results'])
for role in roles:
    print(f'\n Collectioning: {role}')
    for  page in range(1,max_pages+1):
        df=fetch_data(role,page)
        if df.empty:
            print(f"No more jobs for {role}")
            break
        print(f"Page {page}: {len(df)} jobs")
        all_dataframes.append(df)
final_df=pd.concat(all_dataframes,ignore_index=True)
print(final_df.shape)
#final_df.to_csv("raw_jobs.csv", index=False)
```

```python
final_df['company']=final_df['company'].apply(lambda x:x.get('display_name') if isinstance(x,dict) else None )
final_df['category']=final_df['category'].apply(lambda x:x.get('label') if isinstance(x,dict) else None)
final_df['location']=final_df['location'].apply(lambda x:x.get('display_name') if isinstance(x,dict) else None)
```
