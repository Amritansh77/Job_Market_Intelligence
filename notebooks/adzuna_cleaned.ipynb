```python
def standardize_role(title):
    title = str(title).lower().strip()

    # Software Engineer
    if "software engineer" in title:
        return "Software Engineer"

    # Data Scientist
    elif "data scientist" in title or "data science" in title:
        return "Data Scientist"

    # Data Engineer
    elif "data engineer" in title:
        return "Data Engineer"

    # Data Analyst
    elif "data analyst" in title:
        return "Data Analyst"

    # Machine Learning Engineer
    elif "machine learning" in title or "ml engineer" in title:
        return "Machine Learning Engineer"

    # AI Engineer
    elif (
        "artificial intelligence" in title
        or "ai engineer" in title
        or "gen ai" in title
        or "genai" in title
    ):
        return "AI Engineer"

    # Python Developer
    elif "python developer" in title or "python development" in title:
        return "Python Developer"

    # Java Developer
    elif "java developer" in title or "java backend developer" in title:
        return "Java Developer"

    # Backend Developer
    elif "backend developer" in title:
        return "Backend Developer"

    # Frontend Developer
    elif "frontend developer" in title:
        return "Frontend Developer"

    # Full Stack Developer
    elif "full stack" in title or "full-stack" in title:
        return "Full Stack Developer"

    # DevOps Engineer
    elif "devops" in title:
        return "DevOps Engineer"

    # Cloud Engineer
    elif "cloud" in title:
        return "Cloud Engineer"

    # Cyber Security
    elif (
        "cyber security" in title
        or "cybersecurity" in title
        or "cyber" in title
    ):
        return "Cyber Security"

    # Business Analyst
    elif "business analyst" in title or "business analytics" in title:
        return "Business Analyst"

    # Project Manager
    elif "project manager" in title:
        return "Project Manager"

    # Professor / Faculty
    elif (
        "professor" in title
        or "faculty" in title
        or "instructor" in title
    ):
        return "Academic"

    # Sales
    elif "sales" in title:
        return "Sales"
    # Software Development Engineer
    elif "software development engineer" in title:
        return "Software Development Engineer"

    # Platform Engineer
    elif "platform engineer" in title:
        return "Platform Engineer"

    # Software Architect
    elif "software architect" in title:
        return "Software Architect"

    # AI Architect
    elif "ai architect" in title:
        return "AI Architect"

    # Engineering Manager
    elif "engineering manager" in title:
        return "Engineering Manager"

    # Technical Lead
    elif "tech lead" in title or "technical lead" in title:
        return "Technical Lead"

    else:
        return "Other"
```
