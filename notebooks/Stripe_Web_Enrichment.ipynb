```python
import re
import time
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.edge.options import Options
from selenium.webdriver.common.by import By
```

```python
stripe_mask = df['absolute_url'].str.contains(
    'stripe.com',
    case=False,
    na=False
)

stripe_indices = df.index[stripe_mask]

print("Stripe rows:", len(stripe_indices))
```

```python
options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Edge(options=options)
```

```python
def extract_stripe_job(url):

    driver.get(url)

    time.sleep(2)

    actual_url = driver.current_url

    result = {
        'location_clean': None,
        'salary_currency': None,
        'salary_min': None,
        'salary_max': None,
        'education': None,
        'page_type': None,
        'actual_url': actual_url
    }

    # Check whether Stripe opened an individual job page
    if '/careers/listing/' not in actual_url:
        result['page_type'] = 'search_or_redirect'
        return result

    result['page_type'] = 'job_listing'

    try:
        page_text = driver.find_element( By.TAG_NAME,"body").text
    except:
        return result

    text = re.sub(r'\n+', '\n', page_text)

    # ---------------------------------------
    # LOCATION
    # ---------------------------------------

    location_match = re.search( r'Office locations?\s*\n(.*?)(?=\nRemote location|\nEmployment type|\nApply)',
        text, re.IGNORECASE | re.DOTALL
    )

    if location_match:

        location = location_match.group(1).strip()

        location = re.sub( r'\s+', ' ', location )

        result['location_clean'] = location

    # ---------------------------------------
    # SALARY
    # ---------------------------------------

    salary_patterns = [
        r'CA\$\s*([\d,]+)\s*-\s*CA\$\s*([\d,]+)',
        r'US\s*base salary.*?\$\s*([\d,]+)\s*-\s*\$\s*([\d,]+)',
        r'\$\s*([\d,]+)\s*-\s*\$\s*([\d,]+)',
        r'€\s*([\d,]+)\s*-\s*€\s*([\d,]+)',
        r'£\s*([\d,]+)\s*-\s*£\s*([\d,]+)'
    ]

    for pattern in salary_patterns:

        salary_match = re.search( pattern,text, re.IGNORECASE | re.DOTALL )

        if salary_match:

            result['salary_min'] = float(
                salary_match.group(1).replace(',', '')
            )

            result['salary_max'] = float(
                salary_match.group(2).replace(',', '')
            )

            matched_text = salary_match.group(0)

            if 'CA$' in matched_text:
                result['salary_currency'] = 'CAD'

            elif '€' in matched_text:
                result['salary_currency'] = 'EUR'

            elif '£' in matched_text:
                result['salary_currency'] = 'GBP'

            else:
                result['salary_currency'] = 'USD'

            break

    # ---------------------------------------
    # EDUCATION
    # ---------------------------------------

    education_patterns = [
    r"((?:Bachelor['’]?s|Master['’]?s|Ph\.?D\.?)[^\n]*)"
    ]

    education_matches = []

    for pattern in education_patterns:

        matches = re.findall( pattern,text, re.IGNORECASE )

        education_matches.extend(matches)

    if education_matches:

        education_matches = list( dict.fromkeys( x.strip()  for x in education_matches ) )

        result['education'] = "; ".join(education_matches[:3])

    return result
```

```python
test_url = "https://stripe.com/jobs/search?gh_jid=8044839"

test_result = extract_stripe_job(test_url)

print(test_result)
```

```python
stripe_indices = df.index[
    df['absolute_url'].str.contains('stripe.com', case=False, na=False)
]

results = []

for count, idx in enumerate(stripe_indices, start=1):

    url = df.loc[idx, 'absolute_url']

    result = extract_stripe_job(url)

    result['df_index'] = idx
    result['source_url'] = url

    results.append(result)

    if count % 10 == 0:
        print(f"Processed {count}/{len(stripe_indices)}")

    time.sleep(1)
```

```python
stripe_results = pd.DataFrame(results)

print(stripe_results.shape)

print(stripe_results['page_type'].value_counts(dropna=False))
```
