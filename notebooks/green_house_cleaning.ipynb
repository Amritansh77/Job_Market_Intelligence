```python
def get_metadata_value(metadata, target):
    try:
        items = ast.literal_eval(metadata)

        for item in items:
            if isinstance(item, dict) and item.get('name') == target:
                return item.get('value')

    except (ValueError, SyntaxError, TypeError):
        return None

    return None
```

```python
def clean_working_model(x):
    if not isinstance(x, list):
        return None

    mapping = {
        '1-In Office': 'In Office',
        '2-Flexible': 'Flexible',
        '3-Fully Remote': 'Fully Remote',
        '3CF-Customer Facing Remote': 'Customer Facing Remote'
    }

    return ', '.join(mapping.get(item, item) for item in x)
```

```python
df['working_model_clean'] = df['working_model'].apply(
    clean_working_model
)
```
